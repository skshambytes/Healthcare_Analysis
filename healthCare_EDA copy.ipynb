```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sqlalchemy import create_engine
```

```python


df= pd.read_csv('Healthcare_EDA_Dataset.csv')
df.head()
```

```python
df['Gender'].fillna(df['Gender'].mode()[0], inplace=True)
df['BloodGroup'].fillna(df['BloodGroup'].mode()[0], inplace=True)
df['HeightCM'].fillna(df['HeightCM'].median() , inplace=True)
df['WeightKG'].fillna(df['WeightKG'].median() , inplace=True)
df['City'].fillna("Unknown", inplace=True)
df['Diagnosis'].fillna("Not Recorded", inplace=True)
df['Insurance'].fillna("Unknown", inplace=True)
df['TreatmentCost'].fillna(df['TreatmentCost'].median() , inplace=True)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["Grp Age"] = df["Age"].apply(lambda x : "Child" if x < 18 else "Adult" if x < 60 else "Senior")
```

```python
df.to_csv('Cleaned_Healthcare_EDA_Dataset.csv', index=False)
```
